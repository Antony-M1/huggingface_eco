## Downloading NLTK Data


The items you mentioned are all datasets and models provided by the NLTK library. Here's a breakdown of each:

1. **`punkt`**: 
   - **Type**: Tokenization model
   - **Purpose**: It is used for sentence and word tokenization. It helps split a text into sentences or words by recognizing sentence boundaries.
   - **Usage**: Loaded using `nltk.download('punkt')`.

2. **`averaged_perceptron_tagger`**:
   - **Type**: Part-of-speech tagging model
   - **Purpose**: This model is used to assign parts of speech (e.g., noun, verb, adjective) to each word in a sentence based on its context.
   - **Usage**: Loaded using `nltk.download('averaged_perceptron_tagger')`.

3. **`wordnet`**:
   - **Type**: Lexical database
   - **Purpose**: WordNet is a large lexical database of English words that helps with synonyms, antonyms, definitions, and relations between words (like hyponyms and hypernyms).
   - **Usage**: Loaded using `nltk.download('wordnet')`.

## Tokenization

```python
text = "Hello world! Welcome to NLTK. It's a great library for NLP."
```

```python
# Sentence Tokenization
sentance_tokenize = sent_tokenize(text)
print(sentance_tokenize)
```

```python
# Word Tokenization
word_tok = word_tokenize(text)
print(word_tok)
```

```python
# Stop Words
stop_words = set(stopwords.words("english"))
filtered_words = [word for word in word_tok if word.lower() not in stop_words]
print("Words: ", len(word_tok), "Filtered: ", len(filtered_words))
```

**Stemming and Lemmatization**

**Stemming :** *to stop something that is increasing or spreading.*


**Lemmatization :**  the process of reducing the different forms of a word to one single form, for example, reducing `builds`, `building`, or `built` to the **lemma** `build`: Lemmatization is the process of grouping inflected forms together as a single base form.

Stemming reduces a word to its root form, while lemmatization converts a word to its base form using a dictionary.

### Stemming

```python
stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in filtered_words]

data = {
    "Filterd Words" : filtered_words,
    "Stemmed Words": stemmed_words
}

df = pd.DataFrame(data)

df.style.set_caption("Stemmer Info").set_properties(**{
    'background-color': 'lightblue',
    'color': 'black',
    'border-color': 'white'
})
```

## Named Entity Recognition (NER)

Named Entity Recognition (NER) is the process of identifying and classifying key entities in a text (like names of people, organizations, locations, etc.).

```python
def named_entity_recognition(text:str):
    
    tokens = word_tokenize(text)
    pos_tags = pos_tag(tokens)
    named_entity = ne_chunk(pos_tags)
    return named_entity
```

```python
text = "Barack Obama was born in Hawaii. He was the 44th President of the United States."
entities = named_entity_recognition(text)

print(entities)
```

## Parsing and Syntactic Analysis

Syntactic parsing involves analyzing the structure of a sentence based on grammatical rules.

`CFG` stands for `Context-Free Grammar`. It is a formal grammar used to define the syntax of languages, including programming languages and natural languages.

### Define a Grammar

You can define a simple context free grammar

```python
simple_cfg_grammar = CFG.fromstring("""
    S -> NP VP
    NP -> Det N | Det N PP
    VP -> V NP | VP PP
    PP -> P NP
    Det -> 'a' | 'the'
    N -> 'dog' | 'cat' | 'park'
    V -> 'chased' | 'saw'
    P -> 'in' | 'on'
""") 
simple_cfg_grammar
```

### Parse a Sentence

Parse a sentance

```python
!pip install supervisor==4.1.0
```

```python
parser = ChartParser(simple_cfg_grammar)

sentence = 'the dog chased a cat in the park'.split()

for tree in parser.parse(sentence):
    print(tree)
    tree.pretty_print()
```

## Building a Simple Text Classifier

you can built a text calsifier using the NLTK  classification capabilities.

The `NaiveBayesClassifier` in NLTK is an implementation of the Naive Bayes algorithm, a simple but effective probabilistic classifier. It’s widely used in natural language processing (NLP) tasks such as spam detection, sentiment analysis, and text classification.

Create a feature sets from a `movie_reviews`

```python
def extract_features(words):
    return {word:True for word in words}
```

```python
len(movie_reviews.fileids(movie_reviews.categories()[0]))
```

```python
documents = [
    (list(movie_reviews.words(file_id)), category)
    for category in movie_reviews.categories()
    for file_id in movie_reviews.fileids(category)
]

random.shuffle(documents)

feature_sets = [(extract_features(words), category) for words, category in documents]

train_set, test_set = feature_sets[:1500], feature_sets[1500:]

print(f"Traing Set : {len(train_set)}, Test Set : {len(test_set)}")
```

### Train the Classifier

Now train the Naive Byase Classifier

```python
classifier = NaiveBayesClassifier.train(train_set)
classifier
```

```python
accuracy = nltk.classify.accuracy(classifier, test_set)
print(f'Accuracy: {accuracy * 100:.2f}%')
```

## Sentiment Analysis

Perform a sentiment analysis from the trained classifier from the above step

### Predict Sentiment

Use the classifier to predict the sentiment on new text

```python
new_review = "This movie was fantastic! I loved it."
new_feature_set = extract_features(new_review)
sentiment = classifier.classify(new_feature_set)
print(f'Sentiment: {sentiment}')
```
